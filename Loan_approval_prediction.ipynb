```python
# import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
df=pd.read_csv("/content/loan_approval_dataset.csv")
```

```python
df.columns=df.columns.str.strip()
df.columns
```

```python
# check loan staus
print(df["loan_status"].value_counts())
```

```python
# now converted to numerical form
df["loan_status"]=df["loan_status"].str.strip().str.lower()

df["loan_status"]=df["loan_status"].map({
    "approved":1,
    "rejected":0
})
```

```python
print(df["loan_status"].value_counts())
```

```python
# convert categorical columns
# education and self_employed can be text values
df['education']=df['education'].str.strip().str.lower()
df["education"]=df["education"].map({
    "graduate":1,
    "not graduate":0
})
```

```python
df["self_employed"]=df["self_employed"].str.strip().str.lower()
df["self_employed"]=df["self_employed"].map({
    "yes":1,
    "no":0
})
```

```python
# create x and y
x=df[
    [
        "no_of_dependents",
        "education",
        "self_employed",
        "income_annum",
        "loan_amount",
        "loan_term",
        "cibil_score",
        "residential_assets_value",
        "commercial_assets_value"
    ]
]
y=df["loan_status"]
```

```python
# train test split
x_train,x_test,y_train,y_test=train_test_split(
x,
y,
test_size=0.2,
random_state=42,
stratify=y
  )
```

```python
# feature scaling
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
# create logistic regression model
model=LogisticRegression(max_iter=1000)
```

```python
# train:
# now model learns the relationship from training
model.fit(x_train_scaled,y_train)
```

```python
# check model coefficients
print("Intercept:",model.intercept_)
```

```python
print("Coefficients:",model.coef_)
```

```python
# better readable form
coefficients=pd.DataFrame({
    "feature":x.columns,
    "coefficient":model.coef_[0]
})

print(coefficients)
```

```python
# make predictions
y_pred=model.predict(x_test_scaled)
print(y_pred)
```

```python
# Get approval probability
y_probability=model.predict_proba(x_test_scaled)
(y_probability)
```

```python
# so
approved_probability= model.predict_proba(x_test_scaled)[:,1]
```

```python
# create prediction table
results=pd.DataFrame({
    "actual_status":y_test.values,
    "predicted_status":y_pred,
    "Approval probability":approved_probability
})

print(results.head(10))
```

```python
# for percentage
results["Approval probability(%)"]=(
    results["Approval probability"]*100
)

print(results.head(10))
```

```python
# calculate accuracy
accuracy=accuracy_score(y_test,y_pred)
print("Accuracy:",accuracy)
print("Accuracy(%):",accuracy*100)
```

```python
# confusion matrix
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
# visualize
plt.figure(figsize=(6,4))
plt.imshow(cm)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.xticks([0,1],["Rejected","Approved"])
plt.yticks([0,1],["Rejected","Approved"])
for i in range(2):
  for j in range(2):
    plt.text(j,i,cm[i,j],
             ha="center",
             va="center")
plt.show()
```

```python
# classification report
print(classification_report(
     y_test,
     y_pred,
     target_names=["Rejected","Approved"]
     ))
```
